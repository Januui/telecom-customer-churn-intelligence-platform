# Categorical Feature Encoding

## Objective
The objective of this notebook is to convert categorical variables into numerical representations suitable for machine learning models. The encoded dataset will be used for model training and evaluation.


```python
# Import Libraries

import pandas as pd
```

```python
# Load feature engineered dataset

df=pd.read_csv("../data/processed/feature_engineered_telco_churn.csv")
df.head()
```

```python
# Dataset information

df.info()
```

```python
# Inspect categorical values before encoding
for col in df.select_dtypes(include="object").columns:
    print(col)
    print(df[col].unique())
    print("-" * 30)
```

```python
df.nunique()
```

### Observation

The dataset contains both binary and multi-category categorical features. Binary features will be encoded using label mapping, while multi-category features will be transformed using one-hot encoding.

```python
# Encode Target Variable

df['Churn'] = df['Churn'].map({
    "Yes":1,
    "No":0,
})
df["Churn"].value_counts()
```

### Observation

The target variable (`Churn`) was converted from categorical values ("Yes" and "No") to binary values (1 and 0). This format is required because machine learning models operate on numerical target labels.

<!-- ### Observation

The target varibale was converted from text labels ("Yes" and "No") into binary numerical values (1 and 0).  -->

```python
## Encode Binary Values

binary_columns =[
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling"
]

for col in binary_columns:
    df[col] = df[col].map({
        'Yes':1,
        "No":0,
        "Male":1,
        "Female":0
    })

   
```

```python
 df[binary_columns].head()
```

### Observation

Binary categorical features such as `gender`, `Partner`, `Dependents`, `PhoneService`, and `PaperlessBilling` were encoded into numerical values. This transformation saves the original information while making these features suitable for machine learning algorithms.

<!-- ### Observation

Binary categorical variables were converted into numerical values while preserving their original meaning. -->

```python
# One-Hot Encode multi-class features

multi_columns =["MultipleLines",
                "InternetService",
                "OnlineSecurity",
                "OnlineBackup",
                "DeviceProtection",
                "TechSupport",
                "StreamingTV",
                "StreamingMovies",
                "Contract",
                "PaymentMethod",
                "TenureGroup",
                "MonthlyChargeCategory"]
df= pd.get_dummies(
    df,
    columns=multi_columns,
    dtype=int
)

df.head()
```

<!-- ### Observation

Multi-category variables were transformed into dummy variables using one-hot encoding. The first category was dropped to reduce redundancy and avoid multicollinearity. -->


```python
# Verify Encoded Dataset

df.info()
```

```python
df.shape
```

### Observation

Multi-category categorical features were transformed into dummy variables using one-hot encoding. The `drop_first=True` parameter was used to remove one category from each feature, reducing redundancy and helping prevent multicollinearity. The dataset is now fully numerical and ready for machine learning.
