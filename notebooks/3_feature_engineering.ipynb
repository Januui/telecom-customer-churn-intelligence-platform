# Feature Engineering

 The obejective of this notebook is to create meaningful features from the cleaned telecom customer churn dataset. These engineered features help capture customer behaviour more effectively and improve prediction in machine learning models.

```python
# import libraries

import pandas as pd
import numpy as np
```

```python
# load processed dataset

df = pd.read_csv("../data/processed/cleaned_telco_churn.csv")
```

```python
# dataset informtion
df.shape
df.info()
```

### Observation

The processed dataset contains cleaned customer records with no missing values or duplicate entries. This dataset is now ready for feature engineering.

```python
# create TotalServices Feature

service_columns=[
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "DeviceProtection",
    "TechSupport",
    "OnlineBackup",
    "DeviceProtection",
    "StreamingTV",
    "StreamingMovies"
]

df["TotalServices"]=(
    df[service_columns]
    .replace({
        "Yes":1,
        "No":0,
        "No internet service":0,
        "No phone service":0,
        "DSL":1,
        "Fiber optic":1
    })
    .sum(axis=1)
)
```

```python
df['TotalServices'].head()
```

### Observation

The TotalSevices feature represents the total number of telecom services subscribed to by each customer. This feature may help identify whether customers with more subscribed services are less likely to churn.

```python
# create Tenure Feature

df['TenureGroup'] = pd.cut(
    df['tenure'],
    bins=[0,12,24,48,72],
    labels=['0-12','13-24','25-48','49-72']
)
```

```python
df["TenureGroup"].value_counts()
```

### Observation

Customer tenure was grouped into meaningful ranges representing different ages of the customer lifecycle. This may help the model identify churn patterns across new and long term customers.

```python
# create MonethlyChargeCategory Feature

df["MonthlyChargeCategory"] =pd.cut(
    df['MonthlyCharges'],
    bins=[0,35,70,120],
    labels=['Low','Medium','High']
)
```

```python
df['MonthlyChargeCategory'].value_counts()
```

```python
df[["MonthlyCharges","MonthlyChargeCategory"]].head()
```

### Observation

Monthly charges were categorised into spending groups. Grouping contionous values into categories can simplify customer segementation and improve model interpretability.

```python
# create LongTermCustomer Feature
#np.where(condition, value_if_true, value_if_false)

df['LongTermCustomer']= np.where(df["tenure"]>= 24,1,0)
```

```python
df['LongTermCustomer'].value_counts()
```

### Observation

Customers with a tenure of 24 months or more were labelled as long-term customers. This faeture helps to distinguish loyal customers from new customers.

### Observtion

The dataset now contains additional engineered features derived from existing customer information. These features provide richer information for machine learning models.
